```python
import sys
from pathlib import Path
import pandas as pd
from tqdm import tqdm

# Add project root to Python path
PROJECT_ROOT = Path("../").resolve()
sys.path.append(str(PROJECT_ROOT))

from src.config import TRAIN_PATH_STAGE1, TEST_PATH_STAGE1, CRAWLED_ABSTRACT_PATH
from src.utils.utils import load_csv, save_csv
from src.crawler.fallback_crawler import FallbackCrawler

print("Libraries imported successfully.")
```

```python
print("===== LOAD DATA =====")

train_df = load_csv(TRAIN_PATH_STAGE1)
test_df = load_csv(TEST_PATH_STAGE1)

print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)

print("\n===== COLUMNS =====")
print("Train columns:", list(train_df.columns))
print("Test columns:", list(test_df.columns))

display(train_df.head(3))
display(test_df.head(3))
```

```python
print("===== INIT CRAWLER =====")

crawler = FallbackCrawler(timeout=30)

print("Crawler initialized successfully.")
```

```python
print("===== PREPARE CRAWL STORAGE =====")

crawl_results = []

print("Train size:", len(train_df))
print("Test size:", len(test_df))

sample_doi = train_df.iloc[0]["doi"]
sample_id = train_df.iloc[0]["id"]

print("Sample DOI:", sample_doi)
print("Sample ID:", sample_id)
```

```python
print("===== MERGE TRAIN + TEST FOR CRAWLING =====")

all_df = pd.concat([train_df, test_df], axis=0)

print("Total to crawl:", len(all_df))

crawl_results = []

for idx, row in tqdm(all_df.iterrows(), total=len(all_df)):
    doi = row["doi"]
    paper_id = row["id"]

    try:
        result = crawler.fetch_by_doi(
            doi=doi,
            paper_id=paper_id
        )

        if result:
            crawl_results.append(result)

    except Exception as e:
        print(f"[ERROR] id={paper_id} | {str(e)}")

print("===== CRAWL DONE =====")
print("Total crawled:", len(crawl_results))
```

```python
import pandas as pd
import re
import numpy as np

print("===== STRONG CLEAN CRAWL RESULTS =====")

crawl_df = pd.DataFrame(crawl_results)

def deep_clean(x):
    if x is None:
        return ""
    
    if isinstance(x, float) and np.isnan(x):
        return ""
    
    x = str(x)

    x = x.replace("\r\n", " ")
    x = x.replace("\n", " ")
    x = x.replace("\r", " ")
    x = x.replace("¶", " ")
    x = x.replace("\u00b6", "")

    x = x.replace("\t", " ")

    x = re.sub(r"\s+", " ", x)

    return x.strip()

for col in crawl_df.columns:
    crawl_df[col] = crawl_df[col].apply(deep_clean)
```

```python
print("===== SAVE GLOBAL CRAWL DATA =====")

crawl_df = pd.DataFrame(crawl_results)

print("Shape:", crawl_df.shape)
display(crawl_df.head())

save_csv(crawl_df, CRAWLED_ABSTRACT_PATH)

print("Saved to:", CRAWLED_ABSTRACT_PATH)
```
